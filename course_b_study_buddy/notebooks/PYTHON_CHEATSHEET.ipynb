## 3. Strings

````python
name, count = "Ada", 3

print(f"{name} has {count} notes")      # ← the f is what makes {} work
print("{name} has {count} notes")       # ← no f: prints the braces literally
````

````python
# Formatting inside f-strings - used all over this project's output

print(f"{3.14159:.2f}")        # 3.14          2 decimal places
print(f"{1234567:,}")          # 1,234,567     thousands separators
print(f"{name:12}|")           # pad to width 12, left-aligned
print(f"{name:>12}|")          # pad to width 12, right-aligned
print(f"{0.875:.1%}")          # 87.5%
````

````python
# The string methods you'll actually use

s = "  Hello World  "

print(repr(s.strip()))            # remove surrounding whitespace
print(s.lower())
print(s.replace("o", "0"))
print(s.split())                  # ["Hello", "World"]
print("World" in s)               # True - substring test
print(s.strip().startswith("He")) # True
````

````python
# Long strings: two strings next to each other join automatically.
# This is how we keep long prompts readable.

message = (
    "This is one long sentence "
    "written across two lines "
    "with no plus signs anywhere."
)
print(message)
````

## 4. Lists

JavaScript calls these arrays. Same brackets, same zero-indexing.

````python
items = ["a", "b", "c"]

print(items[0])       # "a"   first
print(items[-1])      # "c"   LAST - negative counts from the end
print(items[:2])      # first two
print(items[1:])      # everything from index 1
print(len(items))     # 3     ← len(x), not x.length
````

````python
items = ["a", "b", "c"]

items.append("d")            # push
print(items)

print(sorted(["c", "a", "b"]))    # returns a NEW sorted list
print("b" in items)               # True
````

````python
# Looping with the index too - we use this for progress messages

for i, item in enumerate(["first", "second", "third"], start=1):
    print(f"[{i}] {item}")
````

````python
# A set removes duplicates. We use this to list unique source files.

files = ["a.md", "b.md", "a.md", "a.md"]

print(set(files))            # unordered, duplicates gone
print(sorted(set(files)))    # back to a list, in a predictable order
````

## 5. Dictionaries

JavaScript calls these objects. **Two real differences: keys need quotes, and there's no
dot access.**

````python
student = {"name": "Ada", "score": 9}

print(student["name"])              # ✅ this is how you read a value
# print(student.name)               # ❌ AttributeError - uncomment to see
````

````python
student = {"name": "Ada", "score": 9}

print(student.get("missing"))          # None - no crash
print(student.get("missing", 0))       # 0    - your own default

# Compare: this one WOULD crash
try:
    student["missing"]
except KeyError as error:
    print(f"KeyError: {error}  ← this is why .get() exists")
````

````python
student = {"name": "Ada", "score": 9}

student["year"] = 2          # add
student["score"] = 10        # update

for key, value in student.items():
    print(f"{key:8} = {value}")
````

## 6. Ifs and loops

The braces are gone. **A colon opens the block, and the indentation IS the block.**

In JavaScript, indenting neatly is good manners. In Python it's the actual syntax — so
you can never have code that *looks* like it's inside an `if` but isn't.

````python
score = 7

if score >= 9:
    print("excellent")
elif score >= 5:          # ← "else if" is spelled "elif"
    print("fine")
else:
    print("needs work")
````

````python
# and / or / not - words, not symbols

age, has_notes = 16, True

if age >= 13 and has_notes:
    print("ready")

if not has_notes or age < 13:
    print("not ready")
````

````python
for word in ["one", "two", "three"]:
    print(word)

print("---")

for i in range(3):            # 0, 1, 2
    print(i)

print("---")

for i in range(1, 4):         # 1, 2, 3
    print(i)
````

## 7. Comprehensions

`.map()` and `.filter()` collapse into one piece of syntax. You'll see this constantly.

````python
nums = [1, 5, 3, 9, 7]

print([n * 10 for n in nums])                 # map
print([n for n in nums if n > 4])             # filter
print([n * 10 for n in nums if n > 4])        # both
````

````python
# Read it from the inside out:
#   "for each n in nums, if n > 4, give me n * 10"
#
# Same thing written as a loop, so you can see they're identical:

result = []
for n in nums:
    if n > 4:
        result.append(n * 10)
print(result)
````

````python
# Building a string from a list - this exact line is in assistant.py

docs = [{"file": "a.md", "text": "first"}, {"file": "b.md", "text": "second"}]

print("\n\n---\n\n".join(d["text"] for d in docs))
````

## 8. Functions

````python
def greet(name):
    """This text is a docstring - the function explaining itself.

    We write one for every function in this project. It says WHY the function
    exists and what you get back; the code already says how.
    """
    return f"Hello {name}!"
````

Running that printed nothing. **Defining** a function and **calling** it are different
things — you taught Python a word, you haven't said it yet.

````python
print(greet("TUMO"))
print(greet.__doc__.split("\n")[0])      # the docstring is readable at runtime
````

````python
# Default arguments

def describe(subject, style="short", excited=False):
    ending = "!" if excited else "."
    return f"A {style} description of {subject}{ending}"


print(describe("RAG"))
print(describe("RAG", excited=True))                 # skip ahead by naming it
print(describe("RAG", style="long", excited=True))
````

````python
# Returning two things at once - answer_question() does exactly this

def answer_question(question):
    """Returns the answer AND the sources it used."""
    return "The answer is 42.", ["a.md", "b.md"]


text, sources = answer_question("anything")      # unpacked into two variables
print(text)
print(sources)
````

<div style="border-left: 6px solid #900; background: #fff4f4; padding: 12px 16px; margin: 12px 0;">
<h3 style="color:#900; margin-top:0;">🛑 The famous Python trap</h3>
<p style="color:#900; margin-bottom:0;">
<b>Never use a list (or dict) as a default argument.</b> It is created <b>once</b>, when
the function is defined, and then <i>shared by every call</i>.<br/><br/>
Run the next cell. It is genuinely surprising the first time.
</p>
</div>

````python
# ❌ THE BUG

def add_item_broken(item, basket=[]):
    basket.append(item)
    return basket


print(add_item_broken("apple"))     # ['apple']
print(add_item_broken("pear"))      # ['apple', 'pear']  ← where did the apple come from?!
print(add_item_broken("plum"))      # ['apple', 'pear', 'plum']
````

````python
# ✅ THE FIX - use None, and make a fresh one inside

def add_item(item, basket=None):
    if basket is None:
        basket = []
    basket.append(item)
    return basket


print(add_item("apple"))
print(add_item("pear"))
print(add_item("plum"))
````

You'll meet this exact pattern in `assistant.py`:

```python
def answer_question(question, history=None, k=config.RETRIEVE_K):
    if history is None:
        history = []
```

Without it, the assistant would quietly remember every conversation it had ever had.

## 9. Files and paths

````python
from pathlib import Path

# Path() makes a location object. The / operator joins parts,
# and it works on Windows, Mac and Linux without changing anything.

p = Path("notes") / "biology" / "cells.md"
print(p)
print(p.name)        # cells.md
print(p.stem)        # cells
print(p.suffix)      # .md
print(p.parent)      # notes/biology
````

````python
# Writing and reading. ALWAYS pass encoding="utf-8" -
# without it, Windows mangles every Armenian (or accented) character.

import tempfile

folder = Path(tempfile.mkdtemp())        # a scratch folder, safe to play in
note = folder / "test.md"

note.write_text("Բարև! Hello!", encoding="utf-8")
print(note.read_text(encoding="utf-8"))
````

````python
# Finding files. ** means "including sub-folders".
# sorted() so the order is the same on every machine - this matters more than it sounds.

(folder / "a.md").write_text("a", encoding="utf-8")
(folder / "b.md").write_text("b", encoding="utf-8")
(folder / "notes.txt").write_text("c", encoding="utf-8")

print([p.name for p in sorted(folder.glob("**/*.md"))])
print(folder.exists(), note.exists())
````

````python
# Bytes vs text. A picture or a PDF is bytes; "w" would corrupt it.

(folder / "data.bin").write_bytes(b"\x89PNG not really")
print((folder / "data.bin").read_bytes()[:4])
````

### `if __name__ == "__main__":`

Every runnable file in the project ends with:

```python
def main():
    ...

if __name__ == "__main__":
    main()
```

It means **"only do this if this file was started directly"**.

- `python ingest.py` → the block runs → the index gets built
- `import ingest` from somewhere else → the block does **not** run

That's what lets you reuse a function from a file without accidentally launching the
whole program.

## 11. Reading an error

**Read the LAST line first.** Everything above it is the trail of how Python got there.
The last line is what actually went wrong.

The cells below deliberately cause each common error and catch it, so you can read the
real message without stopping the notebook.

````python
# NameError - by far the most common in notebooks

try:
    print(undefined_thing)
except NameError as e:
    print(f"NameError: {e}")
    print("→ You skipped a cell, or restarted the kernel. Run all cells from the top.")
````

````python
# KeyError

try:
    {"a": 1}["b"]
except KeyError as e:
    print(f"KeyError: {e}")
    print("→ That key isn't in the dictionary. Use .get('b', default) instead.")
````

````python
# IndexError

try:
    [1, 2, 3][10]
except IndexError as e:
    print(f"IndexError: {e}")
    print("→ The list is shorter than you think. Check len() first.")
````

````python
# TypeError on None - usually a function that forgot to return

def forgot_to_return(x):
    result = x * 2          # no return!


try:
    print(forgot_to_return(5) + 1)
except TypeError as e:
    print(f"TypeError: {e}")
    print("→ Something is None. Often a function with no return statement.")
````

````python
# FileNotFoundError

from pathlib import Path

try:
    Path("this_file_does_not_exist.md").read_text(encoding="utf-8")
except FileNotFoundError as e:
    print(f"FileNotFoundError: {e}")
    print("→ Wrong folder or wrong name. Check with Path('x').exists().")
````

### The full lookup table

| Error | Usually means | Fix |
|---|---|---|
| `NameError` | Cell not run, or kernel restarted | Run all cells from the top, in order |
| `ModuleNotFoundError` | Wrong kernel, or venv not activated | Re-pick the kernel; re-activate the venv |
| `IndentationError` | Inconsistent spacing | Never mix tabs and spaces |
| `KeyError` | Key not in the dictionary | `.get(key, default)` |
| `IndexError` | List shorter than you thought | Check `len()` |
| `TypeError: ... NoneType` | Something is `None` | Often a missing `return` |
| `FileNotFoundError` | Wrong path | `Path("x").exists()` |
| `AuthenticationError` / `401` | Bad or missing API key | Check `.env` character by character |

## 12. Patterns you'll meet in this project

Not new syntax — just combinations, so they don't look strange when you hit them.

````python
# A list of (question, answer) pairs - how conversation history is stored

history = [
    ("Why two trackers?", "Because one failed on Flight 4."),
    ("Which is pricier?", "The satellite one."),
]

for question, answer in history:         # unpacked as we loop
    print(f"Q: {question}\n   A: {answer}")
````

````python
# Taking just the last two, and just the first half of each pair

recent = [q for q, _ in history[-2:]]    # the _ means "I don't need this one"
print(recent)
print(" ".join(recent))                  # exactly what build_search_query does
````

````python
# Unique filenames, sorted - exactly what sources_of() does

chunks = [
    {"metadata": {"filename": "hardware.md"}},
    {"metadata": {"filename": "overview.md"}},
    {"metadata": {"filename": "hardware.md"}},
]

print(sorted({c["metadata"]["filename"] for c in chunks}))
````

````python
# A template string filled in later - how the system prompt gets its context

SYSTEM_PROMPT = """
You answer questions using only the notes below.

NOTES:
{context}
"""

print(SYSTEM_PROMPT.format(context="The balloon bursts at 32,000 m."))
````

Notice `.format()` rather than an f-string. The template has to exist **before** we know
what goes in it, so the `{context}` placeholder stays empty until we call `.format()`.

## 13. Survival

### Notebooks

- **Shift + Enter** runs a cell.
- Cells share memory — **run them in order, from the top**.
- A bare value on the last line displays it, no `print` needed. Use this to look inside things.
- Stuck? **Restart kernel → Run All.** Fixes a surprising amount.
- **Every new notebook needs its kernel picked:** top right → Select Kernel → Python
  Environments → the `.venv` one. This is the cause of most "but I installed it!" problems.

### Terminal

```bash
cd study_buddy                     # go into a folder
ls            (Windows: dir)       # list what's there
source .venv/bin/activate          # Mac / Linux  - activate the environment
.venv\Scripts\activate             # Windows
python ingest.py                   # run a file
```

Your prompt shows `(.venv)` when the environment is active. **You must activate it in
every new terminal window.** "It worked yesterday" is almost always this.
