## PART A: The translation table

### 1. Variables

<div style="border-left: 6px solid #06c; background: #f2f7ff; padding: 12px 16px; margin: 12px 0;">
<h3 style="color:#06c; margin-top:0;">🔄 JavaScript → Python</h3>
<pre style="color:#06c; margin-bottom:0; background:none; border:none;">
// JavaScript              # Python
let name = "Ada";          name = "Ada"
const PI = 3.14;           PI = 3.14
let count = 0;             count = 0
</pre>
<p style="color:#06c; margin-bottom:0;">
No <code>let</code>, no <code>const</code>, no semicolons. You just write the name and the
value.<br/><br/>
Python has no real <code>const</code>. Instead there's a convention: if you write a name in
<code>ALL_CAPITALS</code>, everyone reading your code understands "don't change this".
Python won't stop you. Your teammates will.
</p>
</div>

````python
# Try it. Notice how little punctuation there is.

name = "Ada"
age = 36
is_awesome = True

print(name, age, is_awesome)
````

Two things to spot in that output: `True` has a capital T in Python (JavaScript's is
lowercase), and `print` happily takes several values separated by commas.

### 2. Building strings

<div style="border-left: 6px solid #06c; background: #f2f7ff; padding: 12px 16px; margin: 12px 0;">
<h3 style="color:#06c; margin-top:0;">🔄 JavaScript → Python</h3>
<pre style="color:#06c; margin-bottom:0; background:none; border:none;">
// JavaScript                        # Python
`Hello ${name}, you are ${age}`      f"Hello {name}, you are {age}"
</pre>
<p style="color:#06c; margin-bottom:0;">
Backticks become a normal string with an <code>f</code> in front of it — <b>f</b> for
<b>format</b>. And <code>${}</code> loses the dollar sign and becomes just <code>{}</code>.
<br/><br/>
Forgetting the <code>f</code> is the single most common Python typo there is. Without it
you get the literal text <code>Hello {name}</code> and no error at all, which makes it
sneaky.
</p>
</div>

````python
# With the f, and without it. Run this and compare the two lines.

print(f"Hello {name}, you are {age}")
print("Hello {name}, you are {age}")
````

You met a dictionary already, last session: <code>{"type": "image_generation",
"quality": "low"}</code>.
</p>
</div>

### 5. Indentation instead of braces

This is the one that feels strangest, and the one that wins people over fastest.

<div style="border-left: 6px solid #06c; background: #f2f7ff; padding: 12px 16px; margin: 12px 0;">
<h3 style="color:#06c; margin-top:0;">🔄 JavaScript → Python</h3>
<pre style="color:#06c; margin-bottom:0; background:none; border:none;">
// JavaScript                    # Python
if (age > 18) {                  if age > 18:
  console.log("adult");              print("adult")
} else {                         else:
  console.log("minor");              print("minor")
}
</pre>
<p style="color:#06c; margin-bottom:0;">
The braces are gone. A colon opens the block, and <b>the indentation is the block</b>.<br/><br/>
In JavaScript, indenting neatly is good manners. In Python it's the actual syntax — so you
can never have code that looks like it's inside an <code>if</code> but isn't.
</p>
</div>

````python
# Try changing age above and re-running this

if age > 18:
    print("adult")
else:
    print("minor")
````

````python
# Loops. `for (const s of styles)` becomes `for s in styles`

for style in styles:
    print(f"I could draw this in {style}")
````

## PART B: Functions

<div style="border-left: 6px solid #06c; background: #f2f7ff; padding: 12px 16px; margin: 12px 0;">
<h3 style="color:#06c; margin-top:0;">🔄 JavaScript → Python</h3>
<pre style="color:#06c; margin-bottom:0; background:none; border:none;">
// JavaScript                          # Python
function greet(name) {                 def greet(name):
  return `Hello ${name}`;                  return f"Hello {name}"
}
</pre>
<p style="color:#06c; margin-bottom:0;">
<code>function</code> becomes <code>def</code>, the braces become a colon and indentation,
and <code>return</code> is exactly the same.
</p>
</div>

````python
def greet(name):
    return f"Hello {name}!"
````

Notice that running that cell printed nothing. **Defining** a function and **calling** it
are different things. You just taught Python a word; you haven't said it yet.

````python
greet("TUMO")
````

### Default arguments — the good bit

A parameter can have a value it falls back to when you don't supply one. This is how you
make a function that is simple to use for the common case but still adjustable for the
unusual one.

````python
def describe(subject, style="photograph", mood="calm"):
    return f"A {mood} {style} of {subject}"
````

````python
# Only give the one you have to

describe("a lighthouse")
````

````python
# Or override one of them, by name

describe("a lighthouse", mood="stormy")
````

````python
# Or all of them

describe("a lighthouse", style="oil painting", mood="stormy")
````

Look at that middle call again: `describe("a lighthouse", mood="stormy")`.

We skipped `style` entirely and jumped straight to `mood` by naming it. You can't do that
with positional arguments — you'd have to pass something for `style` just to reach `mood`.

This matters more than it looks. It's the difference between a function that's pleasant to
use and one where you have to keep the parameter order memorised.

## PART C: Building `make_image()`

Here is the code from last session, all of it, in one lump:

```python
result = image_model.invoke(prompt)
image_bytes = base64.b64decode(
    [b for b in result.content_blocks if b["type"] == "image"][0]["base64"]
)
display(Image(image_bytes))
```

It works, but you have to read all of it every time to remember what it does. Let's hide it
behind a name.

````python
# Create the image model once, up here, so we don't rebuild it on every call

image_model = init_chat_model(CHAT_MODEL).bind_tools([IMAGE_TOOL])
````

### Step 1: pull out the fiddly part

The ugliest line is the one that digs the image out of the response. Give *that* its own
function first, so it stops being in the way.

````python
def extract_image_bytes(response):
    """
    Find the image inside a model's response and return it as raw bytes.

    The model replies with a list of 'content blocks' - some are text, some are images.
    We want the first image one, and we want its base64 text decoded back into real bytes.
    """
    for block in response.content_blocks:
        if block["type"] == "image":
            return base64.b64decode(block["base64"])
    # If we get here, the model replied without drawing anything at all.
    raise ValueError("The model didn't return an image. What did it say instead?")
````

That triple-quoted text under the `def` is a **docstring**. It's the function's own
explanation of itself, and it's the first thing anyone reads — including you, in three
weeks, when you've forgotten why you wrote this.

The rule we'll follow all course: a docstring says **why** and **what you get back**. The
code already says *how*.

### Step 2: the function we actually wanted

````python
def make_image(prompt, style="", size="square"):
    """
    Generate an image from a text description and return it as bytes.

    prompt - what you want to see, in plain language
    style  - optional extra words describing how it should look
    size   - "square", "wide" or "tall"

    Returns the image as bytes, ready to display or write to a file.
    """
    # Join the subject and the style into one description.
    # The .strip() removes the trailing space you'd get when style is empty.
    full_prompt = f"{prompt}. {style}".strip()

    # Sizes are described in words here so nobody has to remember pixel numbers
    sizes = {"square": "1024x1024", "wide": "1536x1024", "tall": "1024x1536"}

    response = image_model.invoke(
        f"{full_prompt}. Image size: {sizes[size]}."
    )
    return extract_image_bytes(response)
````

### Step 3: use it

````python
# One line. That's the whole point of the last hour.

image_bytes = make_image("a red bicycle leaning against an ivy wall")
display(Image(image_bytes))
````

````python
# Now with a style

image_bytes = make_image(
    "a red bicycle leaning against an ivy wall",
    style="watercolour painting, soft edges, muted colours",
)
display(Image(image_bytes))
````

````python
# And a different shape

image_bytes = make_image("a lighthouse in a storm", style="dramatic, cinematic", size="wide")
display(Image(image_bytes))
````
